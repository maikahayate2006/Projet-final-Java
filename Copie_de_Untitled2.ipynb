```python
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(exclude=np.number).columns.tolist()
print("Variables numériques :", numeric_cols)
print("Variables catégorielles :", categorical_cols)
```

```python
for col in ['Quartier', 'Etat_Logement', 'Parking',
            'Classe_Energie', 'Terrasse']:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
for col in numeric_cols:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False))
```

```python
df_clean = df.copy()
```

```python
df_clean["Quartier"] = df_clean["Quartier"].str.strip()
df_clean["Parking"] = df_clean["Parking"].str.strip()
```

```python
df_clean["Quartier"] = df_clean["Quartier"].replace("Quartier","QG")
```

```python
print(df_clean["Quartier"].value_counts(dropna=False))
print("\nd",df_clean["Parking"].value_counts(dropna=False))
```

```python
n_duplicates = df_clean.duplicated().sum()
print("Nombre de doublons exact:",n_duplicates)
```

```python
df_clean = df_clean.drop_duplicates().copy()
print("Dimensions apres supression des doublons exacts",df_clean.shape)
```
